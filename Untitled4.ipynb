```python
import zipfile
import pandas as pd

# Unzip the file
with zipfile.ZipFile('/content/Tweets.csv.zip', 'r') as zip_ref:
    zip_ref.extractall('/content/')

# Load the CSV file
df = pd.read_csv('/content/Tweets.csv')

# Display the first few rows and dataset info
print("Dataset Shape:", df.shape)
display(df.head())
df.info()
```

```python
import matplotlib.pyplot as plt

# Nos quedamos con el sentimiento y el texto del tuit
df = df[["airline_sentiment", "text"]]
df.columns = ["etiqueta", "texto"]

print(df.head())
print(df["etiqueta"].value_counts())
print(df["etiqueta"].value_counts(normalize=True).round(3))  # porcentajes

df["etiqueta"].value_counts().plot(kind="bar", color=["tomato", "gray", "seagreen"])
plt.title("Sentimiento de los tuits")
plt.xlabel("Sentimiento")
plt.ylabel("Cantidad")
plt.xticks(rotation=0)
plt.show()
```

```python
import re
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, ConfusionMatrixDisplay

# 1. Limpieza básica del texto de los tuits
def limpiar(texto):
    texto = texto.lower()
    texto = re.sub(r"http\S+", "", texto)      # saca links
    texto = re.sub(r"@\w+", "", texto)         # saca menciones (@aerolinea)
    texto = re.sub(r"[^a-z\s]", " ", texto)    # deja solo letras
    texto = re.sub(r"\s+", " ", texto).strip() # saca espacios de más
    return texto

df["texto_limpio"] = df["texto"].apply(limpiar)

# 2. Separamos en entrenamiento y prueba (estratificado)
X_train, X_test, y_train, y_test = train_test_split(
    df["texto_limpio"], df["etiqueta"],
    test_size=0.2, random_state=42, stratify=df["etiqueta"]
)

# 3. TF-IDF con palabras sueltas y pares de palabras (ej: "not good")
vectorizador = TfidfVectorizer(ngram_range=(1, 2), min_df=2)
X_train_vec = vectorizador.fit_transform(X_train)
X_test_vec = vectorizador.transform(X_test)

# 4. Entrenamos y comparamos dos modelos
modelos = {
    "Naive Bayes": MultinomialNB(),
    "Regresión Logística": LogisticRegression(class_weight="balanced", max_iter=1000),
}

for nombre, modelo in modelos.items():
    modelo.fit(X_train_vec, y_train)
    y_pred = modelo.predict(X_test_vec)
    print("=" * 50)
    print(nombre, "- Precisión general:", round(accuracy_score(y_test, y_pred), 4))
    print(classification_report(y_test, y_pred))

# 5. Matriz de confusión de la Regresión Logística
y_pred_lr = modelos["Regresión Logística"].predict(X_test_vec)
ConfusionMatrixDisplay(
    confusion_matrix(y_test, y_pred_lr, labels=["negative", "neutral", "positive"]),
    display_labels=["negative", "neutral", "positive"]
).plot(cmap="Blues")
plt.title("Matriz de confusión - Regresión Logística")
plt.show()
```

```python
ejemplos = [
    "Thanks for the great service, the crew was amazing!",
    "My flight was delayed three hours and nobody told us anything",
    "Does the flight to Boston leave from gate 12?"
]
ejemplos_limpios = [limpiar(t) for t in ejemplos]
print(modelos["Regresión Logística"].predict(vectorizador.transform(ejemplos_limpios)))
```

```python
from sklearn.metrics import recall_score

for nombre, modelo in modelos.items():
    y_pred = modelo.predict(X_test_vec)
    print(nombre)
    print("  Precisión general:", round(accuracy_score(y_test, y_pred), 4))
    recalls = recall_score(y_test, y_pred, labels=["negative", "neutral", "positive"], average=None)
    for etiqueta, r in zip(["negative", "neutral", "positive"], recalls):
        print(f"  Recall {etiqueta}: {round(r, 2)}")
```
